## Part 3 - The Jacobian Matrix for the Constraint Equations

### Why We Need the Jacobian

After linearizing the constraint equations, we arrive at

$$
\mathbf{G}^t \, \Delta \mathbf{q} = \mathbf{r}
$$

where:

- $\Delta \mathbf{q}$ = correction to the force densities
- $\mathbf{r}$ = current constraint residual
- $\mathbf{G}^t$ = Jacobian, or tangent, matrix of the constraints

So the central question becomes:

**How do we actually compute** $\mathbf{G}^t$?

### Definition of the Jacobian

Schek writes

$$
\mathbf{G}^t = \frac{\partial \mathbf{g}^*}{\partial \mathbf{q}}
$$

where:

- $\mathbf{q} \in \mathbb{R}^m$ = vector of force densities
- $\mathbf{g}^*(\mathbf{q}) \in \mathbb{R}^r$ = vector of constraint residuals
- $m$ = number of members
- $r$ = number of constraints

So the Jacobian has size

$$
\mathbf{G}^t \in \mathbb{R}^{r \times m}
$$

Each entry tells us how one constraint changes when one force density changes:
$$
G^t_{kj} = \frac{\partial g_k^*}{\partial q_j}
$$

### Chain Rule Form of the Jacobian

The constraints generally depend on:

- the coordinates $\mathbf{x}, \mathbf{y}, \mathbf{z}$
- the force densities $\mathbf{q}$

So Schek writes the Jacobian using the chain rule:

$$
\mathbf{G}^t
=
\frac{\partial \mathbf{g}^*}{\partial \mathbf{q}}
=
\frac{\partial \mathbf{g}}{\partial \mathbf{x}} \frac{\partial \mathbf{x}}{\partial \mathbf{q}}
+
\frac{\partial \mathbf{g}}{\partial \mathbf{y}} \frac{\partial \mathbf{y}}{\partial \mathbf{q}}
+
\frac{\partial \mathbf{g}}{\partial \mathbf{z}} \frac{\partial \mathbf{z}}{\partial \mathbf{q}}
+
\frac{\partial \mathbf{g}}{\partial \mathbf{q}}
$$

So the main difficulty is computing the coordinate sensitivities:

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}},
\qquad
\frac{\partial \mathbf{y}}{\partial \mathbf{q}},
\qquad
\frac{\partial \mathbf{z}}{\partial \mathbf{q}}
$$

## Part 4 - Implicit Differentiation Instead of Explicit Differentiation

### Why Not Differentiate the Coordinate Solution Directly?

At first, it may seem natural to solve for the coordinates explicitly and then differentiate.

That would mean trying to write something like

$$
\mathbf{x} = \mathbf{x}(\mathbf{q})
$$

and then computing

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}}
$$

directly.

But Schek points out that this becomes difficult in practice.

### What Is Implicit Differentiation?

Implicit differentiation means:

**differentiate an equation that relates the variables, without first solving explicitly for one variable in terms of the others.**

For example, if

$$
x^2 + y^2 = 1
$$

we could solve explicitly for $y$, or we could differentiate the equation directly.

Schek follows the same idea:

- do not try to solve explicitly for the coordinate sensitivities first
- instead, differentiate the **equilibrium equations** directly

### Why This Helps

The coordinates and force densities are coupled through equilibrium.

So rather than working with an explicit formula for the coordinates, Schek works with the equation that must always remain true:

$$
\text{equilibrium} = \text{applied load}
$$

By differentiating that equilibrium equation directly, we can extract how the coordinates change when the force densities change.

### What Is Implicit Differentiation?

Implicit differentiation means:

**differentiate an equation that relates the variables, without first solving explicitly for one variable in terms of the others.**

---

### Example 1 — A Circle

Suppose

$$
x^2 + y^2 = 1
$$

This equation relates $x$ and $y$, but does not isolate one variable.

If we differentiate both sides directly with respect to $x$:

$$
\frac{d}{dx}(x^2 + y^2) = \frac{d}{dx}(1)
$$

we get

$$
2x + 2y \frac{dy}{dx} = 0
$$

so

$$
\frac{dy}{dx} = -\frac{x}{y}
$$

We found the derivative **without first solving explicitly for $y$**.

---

### Example 2 — Compare to Explicit Differentiation

From

$$
x^2 + y^2 = 1
$$

we could solve explicitly for $y$:

$$
y = \sqrt{1 - x^2}
$$

and then differentiate:

$$
\frac{dy}{dx}
=
\frac{d}{dx}\left(\sqrt{1-x^2}\right)
=
-\frac{x}{\sqrt{1-x^2}}
$$

Since

$$
y = \sqrt{1-x^2}
$$

this is the same result as

$$
\frac{dy}{dx} = -\frac{x}{y}
$$

So both approaches work, but **implicit differentiation avoids solving first**.

---

### Example 3 — A Product Relation

Suppose

$$
xq = 10
$$

and we want to know how $x$ changes when $q$ changes.

Differentiate implicitly:

$$
d(xq) = 0
$$

Apply the product rule:

$$
x \, dq + q \, dx = 0
$$

Now solve for the sensitivity:

$$
q \, dx = -x \, dq
$$

$$
\frac{dx}{dq} = -\frac{x}{q}
$$

Again, we found the derivative directly from the equation, without first solving

$$
x = \frac{10}{q}
$$

---

### Why This Matters for Schek

Schek follows the same idea.

He does **not** first try to solve explicitly for

$$
\mathbf{x} = \mathbf{x}(\mathbf{q})
$$

and then differentiate.

Instead, he starts from the equilibrium equation

$$
\mathbf{C}^t \mathbf{U}(\mathbf{x}) \, \mathbf{q} = \mathbf{P}_x
$$

and differentiates that equation directly.

So the strategy is:

- do **not** solve explicitly for the coordinate sensitivities first
- instead, differentiate the **equilibrium equations** directly
- then solve for

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}}
$$

This is exactly the idea of **implicit differentiation**.

## Part 5 - How Schek Sets Up the Implicit Differentiation

### Start From the Equilibrium Equation

For the $x$-coordinates, equilibrium is written as

$$
\mathbf{C}^t \mathbf{U}(\mathbf{x}) \, \mathbf{q} = \mathbf{P}_x
$$

where:

- $\mathbf{C}$ = branch-node connectivity matrix
- $\mathbf{U(x)}$ = diagonal matrix of branch $x$-coordinate differences, obviously it is as a function of $x$-coordinates
- $\mathbf{q}$ = vector of force densities
- $\mathbf{P}_x$ = external nodal loads in the $x$-direction

The key point is that $\mathbf{U}$ depends on the geometry, so it depends on the nodal coordinates.

### View the Left-Hand Side as a Function

Define

$$
\mathbf{F}(\mathbf{x}, \mathbf{q}) = \mathbf{C}^t \mathbf{U}(\mathbf{x}) \, \mathbf{q}
$$

Then equilibrium becomes

$$
\mathbf{F}(\mathbf{x}, \mathbf{q}) = \mathbf{P}_x
$$

This makes the dependence clear:

- changing $\mathbf{q}$ changes equilibrium
- changing $\mathbf{x}$ changes equilibrium through $\mathbf{U}(\mathbf{x})$

So $\mathbf{F}$ depends on both $\mathbf{x}$ and $\mathbf{q}$.

### Perturb the System

Now introduce small changes:

$$
\mathbf{q} \rightarrow \mathbf{q} + d\mathbf{q}
$$

$$
\mathbf{x} \rightarrow \mathbf{x} + d\mathbf{x}
$$

These changes must still preserve equilibrium, so after the perturbation we still require

$$
\mathbf{F}(\mathbf{x} + d\mathbf{x}, \mathbf{q} + d\mathbf{q}) = \mathbf{P}_x
$$

Since we already had

$$
\mathbf{F}(\mathbf{x}, \mathbf{q}) = \mathbf{P}_x
$$

their difference must be zero.

### Take the Total Differential

After perturbing the system, we require that the equilibrium equation is still satisfied.  

So the **total change** in the equilibrium expression must be zero:

$$
d\left(\mathbf{C}^t \mathbf{U} \, \mathbf{q}\right) = \mathbf{0}
$$

Now apply the total differential:

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{q}} \, d\mathbf{q}
+
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}} \, d\mathbf{x}
= \mathbf{0}
$$

This says:

- one part of the change comes from changing the force densities
- the other part comes from changing the geometry
- and these two effects must balance so equilibrium is maintained

### What Does 
$$
d\left(\mathbf{C}^t \mathbf{U} \, \mathbf{q}\right) = \mathbf{0}
$$
Mean?

This expression represents the **total change** in the equilibrium equation.

- $\mathbf{C}^t \mathbf{U}(\mathbf{x}) \mathbf{q}$ is the equilibrium expression
- both $\mathbf{x}$ (through $\mathbf{U}$) and $\mathbf{q}$ can change

So:

$$
d(\cdot)
$$

means:

👉 “take a **small change** in this entire expression due to small changes in $\mathbf{x}$ and $\mathbf{q}$”



Since we require equilibrium to still hold after the change:

$$
\text{equilibrium before} = \text{equilibrium after}
$$

the **total change must be zero**:

$$
d\left(\mathbf{C}^t \mathbf{U} \, \mathbf{q}\right) = \mathbf{0}
$$



### How Do We Expand This?

We now treat

$$
\mathbf{F}(\mathbf{x}, \mathbf{q}) = \mathbf{C}^t \mathbf{U}(\mathbf{x}) \, \mathbf{q}
$$

as a function of **two variables**:

- $\mathbf{x}$
- $\mathbf{q}$



### Total Differential (Multivariable Chain Rule)

For a function of two variables, the total differential is:

$$
d\mathbf{F}
=
\frac{\partial \mathbf{F}}{\partial \mathbf{q}} \, d\mathbf{q}
+
\frac{\partial \mathbf{F}}{\partial \mathbf{x}} \, d\mathbf{x}
$$


### Apply This to the Equilibrium Expression

Substitute back:

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{q}} \, d\mathbf{q}
+
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}} \, d\mathbf{x}
= \mathbf{0}
$$




## Part - Evaluate the Two Partial Derivatives


### Evaluating the Two Partial Derivatives

We now expand

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{q}}
\qquad \text{and} \qquad
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}}
$$

carefully.




### First Partial Derivative: With Respect to $\mathbf{q}$

Consider

$$
\mathbf{C}^t \mathbf{U} \mathbf{q}
$$

When differentiating with respect to $\mathbf{q}$, we treat $\mathbf{C}^t$ and $\mathbf{U}$ as constants.

So this is just a linear expression of the form

$$
\mathbf{A}\mathbf{q}
$$

with

$$
\mathbf{A} = \mathbf{C}^t \mathbf{U}
$$

Therefore,

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{q}} = \mathbf{C}^t \mathbf{U}
$$




### Second Partial Derivative: With Respect to $\mathbf{x}$

This one is more subtle, because $\mathbf{U}$ depends on the geometry.

Write the expression as

$$
\mathbf{C}^t \mathbf{U} \mathbf{q}
$$

Recall that:

- $\mathbf{U} = \operatorname{diag}(\mathbf{u})$
- $\mathbf{Q} = \operatorname{diag}(\mathbf{q})$

Since $\mathbf{U}$ and $\mathbf{Q}$ are both diagonal matrices, they commute:

$$
\mathbf{U}\mathbf{q} = \mathbf{Q}\mathbf{u}
$$

So we can rewrite the equilibrium term as

$$
\mathbf{C}^t \mathbf{U} \mathbf{q}
=
\mathbf{C}^t \mathbf{Q} \mathbf{u}
$$

This is the key identity being used.




### Why Is This Helpful?

Now the dependence on $\mathbf{x}$ is easier to see, because:

- $\mathbf{Q}$ is constant when differentiating with respect to $\mathbf{x}$
- $\mathbf{u}$ depends on $\mathbf{x}$

So:

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}}
=
\frac{\partial (\mathbf{C}^t \mathbf{Q} \mathbf{u})}{\partial \mathbf{x}}
$$

Treating $\mathbf{C}^t$ and $\mathbf{Q}$ as constants gives

$$
\frac{\partial (\mathbf{C}^t \mathbf{Q} \mathbf{u})}{\partial \mathbf{x}}
=
\mathbf{C}^t \mathbf{Q} \frac{\partial \mathbf{u}}{\partial \mathbf{x}}
$$




### What Is $\dfrac{\partial \mathbf{u}}{\partial \mathbf{x}}$?

The vector $\mathbf{u}$ contains the branch coordinate differences in the $x$-direction.

These are obtained from the nodal coordinates using the connectivity matrix:

$$
\mathbf{u} = \mathbf{C}\mathbf{x} + \mathbf{C}_f \mathbf{x}_f
$$

If the fixed coordinates $\mathbf{x}_f$ are constant, then

$$
\frac{\partial \mathbf{u}}{\partial \mathbf{x}} = \mathbf{C}
$$

So the derivative becomes

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}}
=
\mathbf{C}^t \mathbf{Q} \mathbf{C}
$$




### Final Result

So the two derivatives are

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{q}} = \mathbf{C}^t \mathbf{U}
$$

and

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}}
=
\mathbf{C}^t \mathbf{Q} \mathbf{C}
$$

Schek then defines

$$
\mathbf{D} = \mathbf{C}^t \mathbf{Q} \mathbf{C}
$$

so that the second derivative can be written simply as

$$
\frac{\partial (\mathbf{C}^t \mathbf{U} \mathbf{q})}{\partial \mathbf{x}} = \mathbf{D}
$$


## Part 6 - Solving for the Coordinate Sensitivities

### Substitute the Partial Derivatives Back In

Substituting into the total differential equation gives

$$
\mathbf{C}^t \mathbf{U} \, d\mathbf{q} + \mathbf{D} \, d\mathbf{x} = \mathbf{0}
$$

Now rearrange:

$$
\mathbf{D} \, d\mathbf{x} = -\mathbf{C}^t \mathbf{U} \, d\mathbf{q}
$$

### Solve for $d\mathbf{x}$

Multiply both sides by $\mathbf{D}^{-1}$:

$$
d\mathbf{x} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{U} \, d\mathbf{q}
$$

So the coordinate sensitivity matrix is

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{U}
$$

and similarly,

$$
\frac{\partial \mathbf{y}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{V}
$$

$$
\frac{\partial \mathbf{z}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{W}
$$

### Interpretation of the Result

The matrix

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}}
$$

tells us how the nodal $x$-coordinates respond to small changes in the force densities.

Its size is

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}} \in \mathbb{R}^{n \times m}
$$

where:

- $n$ = number of free nodes
- $m$ = number of members

So each column tells us how one force density influences all free-node coordinates in the $x$-direction.

## Part 7 - Forming a Nodal Distance Constraint

### What Is a Nodal Distance Constraint?

Suppose we want the distance between two nodes to match a prescribed value. This is like asking what is the strained length of the member.

For a branch $i$, the current length is

$$
l_i(x,y,z)
$$

and the target length is

$$
l_{u,i}
$$

Then the constraint is written as

$$
g_i(x,y,z) = l_i(x,y,z) - l_{u,i} = 0
$$

So the constraint is satisfied when the current nodal distance equals the prescribed distance.

### Vector Form of the Distance Constraints

If we impose this condition on the first $r$ branches, we can collect them into a vector:

$$
\mathbf{g} = \bar{\mathbf{l}} - \bar{\mathbf{l}}_u
$$

where:

- $\bar{\mathbf{l}}$ = vector of current lengths of the constrained branches (probably shorter than normal vector $\mathbf{l}$)
- $\bar{\mathbf{l}}_u$ = vector of prescribed target lengths

Now we want the Jacobian for this constraint set.

### Partial Derivatives of the Distance Constraint

For the distance constraints, Schek writes

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{x}} = \mathbf{C}^t \bar{\mathbf{U}} \, \bar{\mathbf{L}}^{-1}
$$

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{y}} = \mathbf{C}^t \bar{\mathbf{V}} \, \bar{\mathbf{L}}^{-1}
$$

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{z}} = \mathbf{C}^t \bar{\mathbf{W}} \, \bar{\mathbf{L}}^{-1}
$$

These matrices describe how the constrained branch lengths change with the nodal coordinates.

### Where Do These Derivatives Come From?

We start from the definition of the distance constraint.

For each constrained branch $i$:

$$
g_i = l_i(x,y,z) - l_{u,i}
$$

So the constraint depends on the **branch length**:

$$
l_i = \sqrt{u_i^2 + v_i^2 + w_i^2}
$$

where:

- $u_i$ = $x$-direction difference of the branch  
- $v_i$ = $y$-direction difference  
- $w_i$ = $z$-direction difference  




### Step 1 — Differentiate the Length

Take the derivative of $l_i$ with respect to $x$:

$$
\frac{\partial l_i}{\partial x}
=
\frac{1}{2\sqrt{u_i^2 + v_i^2 + w_i^2}} \cdot 2u_i \frac{\partial u_i}{\partial x}
$$

Simplify:

$$
\frac{\partial l_i}{\partial x}
=
\frac{u_i}{l_i} \frac{\partial u_i}{\partial x}
$$




### Step 2 — Matrix Form

Stacking all branches:

- $\mathbf{l}$ = vector of lengths  
- $\mathbf{u}$ = vector of $x$-differences  

Define:

$$
\bar{\mathbf{L}} = \operatorname{diag}(l_i), \qquad
\bar{\mathbf{U}} = \operatorname{diag}(u_i)
$$

Then:

$$
\frac{\partial \mathbf{l}}{\partial \mathbf{x}}
=
\bar{\mathbf{L}}^{-1} \bar{\mathbf{U}} \frac{\partial \mathbf{u}}{\partial \mathbf{x}}
$$




### Step 3 — Use Connectivity

The vector $\mathbf{u}$ contains the branch coordinate differences in the $x$-direction.

These are obtained from the nodal coordinates using the connectivity matrix:

$$
\mathbf{u} = \mathbf{C}\mathbf{x} + \mathbf{C}_f \mathbf{x}_f
$$

If the fixed coordinates $\mathbf{x}_f$ are constant, then

$$
\frac{\partial \mathbf{u}}{\partial \mathbf{x}} = \mathbf{C}
$$


So:

$$
\frac{\partial \mathbf{u}}{\partial \mathbf{x}} = \mathbf{C}
$$

Substitute:

$$
\frac{\partial \mathbf{l}}{\partial \mathbf{x}}
=
\bar{\mathbf{L}}^{-1} \bar{\mathbf{U}} \mathbf{C}
$$




### Step 4 — Match Schek’s Form

Schek writes the transpose form (row-wise constraints), so:

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{x}}
=
\mathbf{C}^t \bar{\mathbf{U}} \bar{\mathbf{L}}^{-1}
$$




### Same Logic for $y$ and $z$

Using:

- $\mathbf{v} = \mathbf{C} \mathbf{y}$
- $\mathbf{w} = \mathbf{C} \mathbf{z}$

we get:

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{y}} = \mathbf{C}^t \bar{\mathbf{V}} \bar{\mathbf{L}}^{-1}
$$

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{z}} = \mathbf{C}^t \bar{\mathbf{W}} \bar{\mathbf{L}}^{-1}
$$



### Substitute Into the Chain Rule

For this distance constraint, there is no direct dependence on $\mathbf{q}$, so

$$
\frac{\partial \mathbf{g}}{\partial \mathbf{q}} = \mathbf{0}
$$

Therefore the Jacobian becomes

$$
\mathbf{G}_d^t
=
\frac{\partial \mathbf{g}}{\partial \mathbf{x}} \frac{\partial \mathbf{x}}{\partial \mathbf{q}}
+
\frac{\partial \mathbf{g}}{\partial \mathbf{y}} \frac{\partial \mathbf{y}}{\partial \mathbf{q}}
+
\frac{\partial \mathbf{g}}{\partial \mathbf{z}} \frac{\partial \mathbf{z}}{\partial \mathbf{q}}
$$

Now substitute the coordinate sensitivities from the previous part.

### Final Jacobian for Nodal Distance Constraints

Using

$$
\frac{\partial \mathbf{x}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{U},
\qquad
\frac{\partial \mathbf{y}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{V},
\qquad
\frac{\partial \mathbf{z}}{\partial \mathbf{q}} = -\mathbf{D}^{-1} \mathbf{C}^t \mathbf{W}
$$

Schek obtains

$$
\mathbf{G}_d^t
=
-\bar{\mathbf{L}}^{-1}
\left(
\bar{\mathbf{U}}\mathbf{C}\mathbf{D}^{-1}\mathbf{C}^t\mathbf{U}
+
\bar{\mathbf{V}}\mathbf{C}\mathbf{D}^{-1}\mathbf{C}^t\mathbf{V}
+
\bar{\mathbf{W}}\mathbf{C}\mathbf{D}^{-1}\mathbf{C}^t\mathbf{W}
\right)
$$

This matrix tells us how the nodal distance constraints change when the force densities change.

### Main Takeaway

The procedure is:

1. define the constraint function
2. compute its derivatives with respect to the coordinates
3. compute the coordinate sensitivities with respect to the force densities
4. combine them through the chain rule

That is how Schek turns a geometric constraint, such as a prescribed nodal distance, into a Jacobian matrix that can be used in the iterative update.
