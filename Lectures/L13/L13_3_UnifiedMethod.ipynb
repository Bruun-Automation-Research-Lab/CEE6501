## Part 1 - Force Density in the Context of the Unified Framework

### A Reminder Of The Unified Framework

From last lecture

These three method families can be written within a common iterative framework.

A representative expression is:

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\left(\mathbf{C}^{T}\mathbf{K}\mathbf{C}\right)^{-1}
\left(
\mathbf{p}
-
\mathbf{C}^{T}\mathbf{L}^{-1}\mathbf{F}\mathbf{C}_f\,\mathbf{x}_f
-
\mathbf{C}^{T}\mathbf{L}^{-1}\mathbf{F}\mathbf{C}\,\mathbf{x}_i
\right)
$$

The main idea is that all three methods can be viewed as the following process:

1. writing equilibrium on a network of nodes and branches (i.e., members or elements)
2. evaluating the current internal force state
3. updating the nodal coordinates toward equilibrium

## Part 1 - Force Density in the Context of the Unified Framework

### Looking at Terms in the Residual

L^-1*F = what does this remind you of?

In the Schek derivation this was a constant term = Force Density!

So we can write:

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\left(\mathbf{C}^{T}\mathbf{K}\mathbf{C}\right)^{-1}
\left(
\mathbf{p}
-
\mathbf{C}^{T}\mathbf{Q}\mathbf{C}_f\,\mathbf{x}_f
-
\mathbf{C}^{T}\mathbf{Q}\mathbf{C}\,\mathbf{x}_i
\right)
$$

But for constant Q, Schek also defined 

$$
\mathbf{D} = \mathbf{C}^{T}\mathbf{Q}\mathbf{C},
\qquad
\mathbf{D}_f = \mathbf{C}^{T}\mathbf{Q}\mathbf{C}_f
$$

So we can write:

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\left(\mathbf{C}^{T}\mathbf{K}\mathbf{C}\right)^{-1}
\left(
\mathbf{p}
-
\mathbf{D}_f\,\mathbf{x}_f
-
\mathbf{D}\,\mathbf{x}_i
\right)
$$

### The Stiffness Matrix

We never talked about the stiffness term in Schek. 

This is because it us a geometric stiffness term, there was no mention of material in the Schek derivation

This is explained well in the Veenendaal paper, but going from nodal stiffness the we have thought about it, with 3 DOFs per node for elastic truss analysis, or 6x6 for full truss element. 

But what is going on behind the scenes in the Force Density derivation, when thought about in terms of the unified method

The idea is you're effectively "lumping the stiffness" at  the nodes (not thinking about elements). So you get a single term for elastic stiffness and a single term for geometric stiffness. This is called dynamic relaxation, but we won't have time to cover in detail in this class.

in Force Density, this lumped stiffness approach is further simplied, with elastic stiffness being ignored

### Multiplying out the Expression

Start from

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\mathbf{D}^{-1}
\left(
\mathbf{p}
-
\mathbf{D}_f\,\mathbf{x}_f
-
\mathbf{D}\,\mathbf{x}_i
\right)
$$

Distribute \( \mathbf{D}^{-1} \):

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\mathbf{D}^{-1}\mathbf{p}
-
\mathbf{D}^{-1}\mathbf{D}_f\,\mathbf{x}_f
-
\mathbf{D}^{-1}\mathbf{D}\,\mathbf{x}_i
$$

Use \( \mathbf{D}^{-1}\mathbf{D} = \mathbf{I} \):

$$
\mathbf{x}_{i+1}
=
\mathbf{x}_i
+
\mathbf{D}^{-1}\mathbf{p}
-
\mathbf{D}^{-1}\mathbf{D}_f\,\mathbf{x}_f
-
\mathbf{x}_i
$$

Cancel \( \mathbf{x}_i \):

$$
\mathbf{x}_{i+1}
=
\mathbf{D}^{-1}\mathbf{p}
-
\mathbf{D}^{-1}\mathbf{D}_f\,\mathbf{x}_f
$$

So the expanded and simplified form is

$$
\mathbf{x}_{i+1}
=
\mathbf{D}^{-1}
\left(
\mathbf{p}
-
\mathbf{D}_f\,\mathbf{x}_f
\right)
$$

and you can do this for y and z, same process


This is the same expression from the Schek derivation!

But the important thing is that the iterative process reduces down to a linear problem when Q is given

 So, what this is saying is that the nodal positions are calculated as a linear operation. No iteration required when force densities are given, but elegant that this comes out of the overall unified expression.
